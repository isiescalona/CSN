Instrucciones

Descripción del Problema: Imagina que estás trabajando como analista de datos para una empresa que recopila información sobre las ventas diarias de diferentes productos en varias tiendas. Necesitas analizar estos datos para extraer información útil y presentar un resumen de las ventas.

Requisitos del Sistema:
El programa debe generar datos de ventas de manera aleatoria para 10 productos diferentes en 5 tiendas distintas durante 7 días (una semana).
Los datos deben estar almacenados en un array de NumPy de 3 dimensiones (productos, tiendas, días).

Debes calcular y mostrar la siguiente información:
Total de ventas por producto a lo largo de la semana.
Total de ventas por tienda a lo largo de la semana.
Promedio de ventas por producto por día.
Promedio de ventas por tienda por día.
Producto con mayor y menor ventas totales en la semana.
Tienda con mayor y menor ventas totales en la semana.

Detalles Específicos:
Genera los datos de ventas aleatorias usando np.random.randint, con valores entre 0 y 100.
Usa funciones de NumPy para calcular las sumas y promedios.
Maneja posibles errores o valores inesperados.
Presenta los resultados de manera clara y estructurada.

Consideraciones Adicionales:
Organiza tu código utilizando funciones para cada una de las tareas principales.
Usa comentarios para explicar cada sección de tu código.
Asegúrate de que el código sea robusto y maneje casos en los que los datos generados puedan estar fuera de los rangos esperados.

```python
import numpy as np


# Generar ventas aleatorias entre 0 y 100.
def generar_datos_ventas(productos, tiendas, dias):
    for cantidad in (productos, tiendas, dias):
        if type(cantidad) != int or cantidad <= 0:
            raise ValueError("Las cantidades deben ser números enteros positivos.")

    return np.random.randint(0, 101, (productos, tiendas, dias))


# Comprobar que los datos sean válidos.
def validar_datos(datos):
    if not isinstance(datos, np.ndarray):
        raise ValueError("Los datos deben ser un array de NumPy.")

    if datos.ndim != 3 or 0 in datos.shape:
        raise ValueError("El array debe tener tres dimensiones y no estar vacío.")

    if datos.dtype.kind not in "iuf":
        raise ValueError("Las ventas deben ser números reales.")

    if not np.all(np.isfinite(datos)):
        raise ValueError("Los datos contienen valores NaN o infinitos.")

    if np.any(datos < 0) or np.any(datos > 100):
        raise ValueError("Las ventas deben estar entre 0 y 100.")

    if np.any(datos != np.floor(datos)):
        raise ValueError("Las unidades vendidas deben ser números enteros.")


# Sumar tiendas y días para obtener el total de cada producto.
def calcular_totales_ventas_por_producto(datos):
    return np.sum(datos, axis=(1, 2))


# Sumar productos y días para obtener el total de cada tienda.
def calcular_totales_ventas_por_tienda(datos):
    return np.sum(datos, axis=(0, 2))


# Calcular el promedio de cada producto por tienda y día.
def calcular_promedio_ventas_por_producto(datos):
    return np.mean(datos, axis=(1, 2))


# Calcular el promedio de cada tienda por producto y día.
def calcular_promedio_ventas_por_tienda(datos):
    return np.mean(datos, axis=(0, 2))


# Calcular el promedio diario de cada producto, sumando todas las tiendas.
def calcular_promedio_diario_por_producto(datos):
    ventas_por_dia = np.sum(datos, axis=1)
    return np.mean(ventas_por_dia, axis=1)


# Calcular el promedio diario de cada tienda, sumando todos los productos.
def calcular_promedio_diario_por_tienda(datos):
    ventas_por_dia = np.sum(datos, axis=0)
    return np.mean(ventas_por_dia, axis=1)


# Encontrar las posiciones del producto con mayor y menor venta.
def encontrar_producto_mayor_menor_ventas(totales):
    mayor = np.argmax(totales)
    menor = np.argmin(totales)
    return mayor, menor


# Encontrar las posiciones de la tienda con mayor y menor venta.
def encontrar_tienda_mayor_menor_ventas(totales):
    mayor = np.argmax(totales)
    menor = np.argmin(totales)
    return mayor, menor


# Mostrar los resultados del análisis.
def imprimir_resumen(datos):
    validar_datos(datos)

    productos, tiendas, dias = datos.shape

    totales_producto = calcular_totales_ventas_por_producto(datos)
    totales_tienda = calcular_totales_ventas_por_tienda(datos)

    promedios_producto = calcular_promedio_ventas_por_producto(datos)
    promedios_tienda = calcular_promedio_ventas_por_tienda(datos)

    promedios_diarios_producto = calcular_promedio_diario_por_producto(datos)
    promedios_diarios_tienda = calcular_promedio_diario_por_tienda(datos)

    producto_mayor, producto_menor = encontrar_producto_mayor_menor_ventas(
        totales_producto
    )
    tienda_mayor, tienda_menor = encontrar_tienda_mayor_menor_ventas(
        totales_tienda
    )

    print("=" * 60)
    print("ANÁLISIS SEMANAL DE VENTAS")
    print("=" * 60)
    print(f"Productos: {productos}")
    print(f"Tiendas: {tiendas}")
    print(f"Días: {dias}")
    print(f"Forma del array: {datos.shape}")
    print(f"Total general: {np.sum(datos)} unidades")

    print("\nVENTAS POR PRODUCTO")
    print("-" * 60)

    for i in range(productos):
        print(f"\nProducto {i + 1}")
        print(f"  Total semanal: {totales_producto[i]} unidades")
        print(f"  Promedio por tienda y día: {promedios_producto[i]:.2f}")
        print(
            "  Promedio diario sumando todas las tiendas: "
            f"{promedios_diarios_producto[i]:.2f}"
        )

    print("\nVENTAS POR TIENDA")
    print("-" * 60)

    for i in range(tiendas):
        print(f"\nTienda {i + 1}")
        print(f"  Total semanal: {totales_tienda[i]} unidades")
        print(f"  Promedio por producto y día: {promedios_tienda[i]:.2f}")
        print(
            "  Promedio diario sumando todos los productos: "
            f"{promedios_diarios_tienda[i]:.2f}"
        )

    # Sumamos 1 porque las posiciones de Python comienzan en 0.
    print("\nMAYORES Y MENORES VENTAS")
    print("-" * 60)

    print(
        f"Producto con mayor venta: Producto {producto_mayor + 1}, "
        f"con {totales_producto[producto_mayor]} unidades."
    )
    print(
        f"Producto con menor venta: Producto {producto_menor + 1}, "
        f"con {totales_producto[producto_menor]} unidades."
    )
    print(
        f"Tienda con mayor venta: Tienda {tienda_mayor + 1}, "
        f"con {totales_tienda[tienda_mayor]} unidades."
    )
    print(
        f"Tienda con menor venta: Tienda {tienda_menor + 1}, "
        f"con {totales_tienda[tienda_menor]} unidades."
    )

    print("\nSi hay empate, se muestra el primero que aparece.")

    # Ejemplos de indexación y slicing.
    print("\nCONSULTAS DE VENTAS")
    print("-" * 60)
    print("Producto 1, tienda 1, día 1:", datos[0, 0, 0])
    print("Producto 1, tienda 1, toda la semana:", datos[0, 0, :])
    print("Producto 1, tienda 1, primeros tres días:", datos[0, 0, :3])

# 1. Generar los datos de ventas para 10 productos, 5 tiendas y 7 días
datos_ventas = generar_datos_ventas(productos=10, tiendas=5, dias=7)

# 2. Ejecutar el análisis y mostrar los resultados impresos
imprimir_resumen(datos_ventas)
```
